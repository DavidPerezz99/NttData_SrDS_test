## Datos relacionales (SQLite)

````python
import sqlite3
import pandas as pd

con = sqlite3.connect('datos_conecta/conecta.db')
pd.read_sql("SELECT name FROM sqlite_master WHERE type='table'", con)
````

````python
# Ejemplo de consulta SQL
pd.read_sql('SELECT * FROM clientes LIMIT 5', con)
````

Si prefieres DuckDB (gratuito, fácil de instalar en Colab), también puede leer el archivo SQLite:

## Datos no relacionales (documentos JSON)

````python
import json, gzip

with open('datos_conecta/tickets.jsonl', encoding='utf-8') as f:
    tickets = [json.loads(linea) for linea in f]
print(len(tickets), 'tickets')
tickets[0]
````

````python
with gzip.open('datos/eventos.jsonl.gz', 'rt', encoding='utf-8') as f:
    eventos = [json.loads(linea) for linea in f]
print(len(eventos), 'eventos')
eventos[0]
````
